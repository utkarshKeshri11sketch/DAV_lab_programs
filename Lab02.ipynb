```python
import numpy as np
import pandas as pd
```

```python
data={
    'Name':['Alice','Bob','Utkarsh','Utkarsh','Suyash','Sushant','Upendra','Samarth','Alice'],
    'Age':[25,88,27,np.nan,88,np.nan,92,8,25],
    'Marks':[92,73,66,33,np.nan,93,np.nan,88,92],
    'City':[np.nan,'Mumbai',np.nan,'Mumbai','Delhi','Mumbai','Delhi',np.nan,np.nan]
    }
df=pd.DataFrame(data)
print("Dataset:")
print(df)
```

```python
print("\nShape of Dataset:")
print(df.shape)
```

```python
print("\nColumn names:",df.columns.tolist())
```

```python
print("\nData types of each column:")
print(df.dtypes)
```

```python
print("\nDataset info:")
df.info()
```

```python
print("\nDescriptive statistics (numerical columns):")
print(df.describe())
```

```python
print("\nMean:\n",df.mean(numeric_only=True))
```

```python
print("\nMedian:\n",df.median(numeric_only=True))
```

```python
print("\nStandard Deviation:\n",df.std(numeric_only=True))
```

```python
print("\nMissing values in each column:")
print(df.isnull().sum())
print("\nTotal missing values in dataset:",df.isnull().sum().sum())
```

```python
print("\nDublicate rows in dataset:")
print(df[df.duplicated()])
print("\nNumber of dublicate rows:", df.duplicated().sum())
print("\nEDA process completed successfully!")
```
